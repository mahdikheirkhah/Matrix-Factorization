```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load factors and movie data
U = np.load("reports/pmf_factors/U_factors_best.npy")
V = np.load("reports/pmf_factors/V_factors_best.npy")
movies = pd.read_csv("data/movies.dat", sep="::", engine="python", names=['movie_id', 'title', 'genres'], encoding="ISO-8859-1")

print("--- 1. GLOBAL INTERPRETABILITY ---")
for factor in range(3):
    top_indices = np.argsort(V[:, factor])[-5:][::-1]
    top_movies = movies.iloc[top_indices]['title'].values
    print(f"Latent Factor {factor} seems to represent these types of movies:")
    for m in top_movies:
        print(f"  - {m}")
    print()

print("--- 2. LOCAL INTERPRETABILITY ---")
user_idx = 0  # First user
movie_idx = np.argmax(np.dot(U[user_idx], V.T)) # Their top predicted movie
contributions = U[user_idx] * V[movie_idx]
top_factor = np.argmax(np.abs(contributions))
print(f"Why was '{movies.iloc[movie_idx]['title']}' recommended to User ID 1?")
print(f"Because the strongest driver was Latent Factor {top_factor}, where both the user and movie share high weights.\n")

print("--- 3. MATRIX VISUALIZATION ---")
plt.figure(figsize=(10, 4))
sns.heatmap(V[:20, :15], cmap="RdBu_r", center=0)
plt.title("Latent Factors Heatmap (First 20 Movies, First 15 Factors)")
plt.xlabel("Latent Factors")
plt.ylabel("Movie Index")
plt.show()
```
